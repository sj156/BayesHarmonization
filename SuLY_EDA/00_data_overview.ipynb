# SuLY-MPRAGE: Data Overview

## 1. Scans and subjects

**276 scans, 17 subjects.**

Every `.nii.gz` in the handoff is one MPRAGE scan; there is nothing else in
the tree.

The design matrix `presence_matrix.csv` lists **19** subjects. Two were not
delivered: `sub-0010` (78 scans) and `sub-0004` (no MPRAGE logged).

## 2. Parameter settings

**5 settings: `Param01` - `Param05`.**

Five MPRAGE protocol variants, run on the same subject within one session.

Not every scanner ran all five. `Param03` is absent on Philips and UIH -
a structural gap, not missing data. The full 5-setting grid exists only on
GE, Vida and Prisma; including Philips or UIH leaves 4 settings.

What each setting actually is (TR / TE / TI / flip angle / acceleration) is
not recoverable from the handoff - there are no sidecar files, only the 276
`.nii.gz`. We also cannot confirm that `Param01` denotes the same protocol
across vendors.
